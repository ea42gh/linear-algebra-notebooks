# 3. Specify a Matrix Directly

##### **Specify a Matrix**

```python
def set_A(A, name="ggbApp" ):
    '''set the transformation matrix: it is defined by [u v] where the vectors u, v are in polar coordinates'''
    u = A[:,0];
    tu = np.round(np.arctan2(u[1],u[0])* 180/np.pi,decimals=1)
    ru = np.round(np.linalg.norm(u),decimals=1)
    v = A[:,1]
    tv = np.round(np.arctan2(v[1],v[0])*180/np.pi,decimals=1)
    rv = np.round(np.linalg.norm(v),decimals=1)
    cmd = Template(r"""${name}.evalCommand('r_1=${ru}'); ${name}.evalCommand('theta_1=${tu} deg'); ${name}.evalCommand('r_2=${rv}'); ${name}.evalCommand('theta_2=${tv} deg');""")\
            .substitute(ru=ru,tu=tu,rv=rv,tv=tv,name=name)
    display(Javascript(cmd))

set_A(np.array([[2,2],[1,-3]]).T)  # note there are som rounding issues (TODO: fix this!)
```

##### **Specify a Rotation**

```python
def set_rotation( theta_deg, name="ggbApp" ):
    ru = 1.
    rv = 1.
    tu = np.round(theta_deg,    decimals=1)
    tv = np.round(theta_deg+90.,decimals=1)
    cmd = Template('''
        ${name}.evalCommand('r_1=${ru}');  ${name}.evalCommand('theta_1=${tu} deg');
        ${name}.evalCommand('r_2=${rv}');  ${name}.evalCommand('theta_2=${tv} deg');
    ''').substitute(ru=ru,tu=tu,rv=rv,tv=tv,name=name)
    display(Javascript(cmd))

pn.interact( set_rotation, theta_deg=pn.widgets.IntSlider( start=-360, end=360, value=50 ) )
```

##### **Specify a Reflection**

```python
def set_reflection( theta_deg, name="ggbApp" ):
    c = np.cos( theta_deg * np.pi / 90.)
    s = np.sin( theta_deg * np.pi / 90.)
    A = np.array( [[c,s], [s, -c]])
    set_A( A, name )

set_reflection( 30 )
```
