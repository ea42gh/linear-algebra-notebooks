```python
import numpy as np
import pandas as pd

np.set_printoptions( precision=3 )
import holoviews as hv
hv.extension('bokeh')
import panel as pn
pn.extension()
```

```python
def plot_result( t, y, y_measured, y_hat, error):
    o1 = {'Curve.Theoretical': dict( color='k', line_width=.5, padding=.05),
          'Curve.Y_hat':       dict( color='blue',         muted_alpha=.01 ),
          'Scatter.Y_hat':     dict( color='blue', size=5, muted_alpha=.01 ),
          'Scatter.Measured':  dict( color='red',  size=5, muted_alpha=.01 ),
          'ErrorBars.Measured':dict( lower_head=None, upper_head=None, color='red' ),
          'Overlay':           dict( legend_position = 'bottom_right', width=500)
    }
    h1=\
       hv.Curve  ((t,y     ),'t', 'y', label='Theoretical', group='Theoretical')*\
       hv.Curve  ((t, y_hat),          label='y_hat',       group='Y_hat'      )*\
       hv.Scatter((t, y_hat),          label='y_hat',       group='Y_hat'      )*\
       hv.Scatter((t,y_measured),      label='Measured',    group='Measured'   )*\
       hv.ErrorBars(np.stack([t,y_hat,0*y_hat,y_measured-y_hat], axis=1),
                    kdims='x',vdims=['y', ' ', 'error'],    group='Measured')
    # FIX: Construct Paths rather than ErrorBars

    h2 = \
    hv.Scatter( (t, error), 't', 'error').opts( height=100,width=500, padding=0.05, yticks=2, color='red')*\
    hv.HLine(0)                          .opts(line_width=.5)

    return (h1.opts(o1)+h2).opts(title='Estimated Curve and Estimated Measurement Error').cols(1)
```

# 1. Perform an experiment

Consider an experiment: given an input $t$ (assumed accurate), the experiment produces a value $y_{measured}$.

$\quad\quad\quad t \rightarrow \textbf{Apparatus} \rightarrow y_{measured}$

In the following, we will simulate this experiment by assuming the output should be some function $y = f(t)$<br>
(the theoretical value), and model the measured value by adding some error to this theoretical value.

```python
t          = np.linspace(-2., 3., 30)
y          = np.sin(.6*t)
y_measured = y + .2*np.random.normal(size=30)

df = pd.DataFrame( {'t': t, 'y_theoretical': y, 'y_measured': y_measured})
```

```python
# Plot it
# ===========================================================================
hv.Curve  (df, 't', 'y_theoretical', label='Theoretical').opts(color='k', line_width=.4)*\
hv.Scatter(df, 't', 'y_measured',    label='Measured'   )\
.opts(size=5,color='red', padding=.05, width=500, height=320,title='Experiment: y = sin(.6 t)', legend_position='bottom_right')+\
hv.Table( df[['t', 'y_measured', 'y_theoretical']])
```

> Note that the graph can be thought of as a vector: it represents a vector $y_i$, with indices $i$, and a vector $t_i$. The plot shows $y_i$ versus $t_i$.<br>
Both the $y_{theoretical}$ and $y_{measured}$ values are shown.

# 2. Conjecture a linear model
We assume that the theoretical $y$ values are related to the $t$ values by a function $y = f(t)$<br>
(hopefully based on some theory, rather than by an impression derived from looking at the graph above!)<br>
Here, we assume that the relationship is a straight line.

Substitute our measurement data into the equation for a straight line $y = \alpha + \beta t$<br>
This an equation of the form $A x = b$, where $A = ( 1, t )$, $x = (\alpha \; \beta)^t$ and $b = y_{measured}$.<br>
Note the first column of $A$ multiplies $\alpha$ and hence is all ones.

Even if our model is correct, the $y_{measured}$ values we are substituting contain error:<br>
the resulting system is very unlikely to have a solution!

## 2.1 Solve the resulting Normal Equations

### 2.1.1 Compute and solve the normal equation:
$A^t A x = A^t y$ to get estimates $x^* = \left(\hat{\alpha}, \hat{\beta} \right)$, the unknowns in our model.

```python
A    = np.stack( [np.ones(len(t)), t ],axis=1)
At_A = A.T @ A
At_y = A.T @ y_measured

x_star = np.linalg.solve( At_A, At_y )

print( 'At_A =');    print(At_A); print()
print( 'At_y =');    print(At_y); print()
print('Solution ='); print(x_star)
```

### 2.1.2 Compute the projections

The estimated value $\hat{y} = \hat{\alpha} + \hat{\beta} t$ is the projection of $y_{measured}$ onto the column space $\mathscr{C}(A)$.<br>
The orthogonal component $e = y_{measured} - \hat{y}$ is the change we make to the $y_{measured}$ vector to ensure that our equation $A x = b - e$
has a solution:<br>
it is the shortest vector possible: $\mid\mid e \mid\mid \, = min_x \mid\mid A x - b \mid\mid$  
$\qquad
\begin{aligned}
\hat{y} & = A x^* \\
e       & = y_{measured} - \hat{y}
\end{aligned}
$

The *least mean sqares error* is the average of
$\displaystyle \sum_i e_i^2 = \frac{1}{N}\,\lVert e \rVert^2$, where $N$ is the number of experiments performed,<br>
i.e., the average of the squares of the estimated errors.

```python
# Compute the projections

y_hat = A @ x_star
error = y_measured - y_hat

print( 'Norm(Error)              = ', np.round(np.linalg.norm(error),3))
print( 'Least mean squares error = ', np.round(error.dot( error)/len(error),3) )
plot_result( t, y, y_measured, y_hat, error)
```

**Reminder**
* The measured y value $y_{measured}$ is the $b$ vector in $A x = b$.
* The estimated y value $\hat{y}$ is the orthogonal projection of $b$ onto the column space of $A$, i.e., $b_{_\parallel}$.
* The error $e = y_{measured} - \hat{y}$ is the orthogonal component $b_\perp$.
* The entries of the vectors are plotted as vector entries versus the corresponding t vector entries.

# 3. Conjecture a cubic model
Repeat the above computation with the model $y = \alpha + \beta t + \gamma t^2 + \delta t^3$

```python
t2 = t*t; t3 = t*t*t
A  = np.stack( [np.ones(len(t)), t, t2, t3 ],axis=1)

At_A = A.T @ A
At_y = A.T @ y_measured

x_star = np.linalg.solve( At_A, At_y )

print( 'At_A =');    print(At_A); print()
print( 'At_y =');    print(At_y); print()
print('Solution ='); print(x_star);
```

```python
# Compute the projections

y_hat = A @ x_star
error = y_measured - y_hat

print( 'Norm(Error) =', np.round(np.linalg.norm(error),3))
print( 'Least mean squares error = ', np.round(error.dot( error)/len(error),3) )
print()
plot_result( t, y, y_measured, y_hat, error)
```

# 4. Overfitting and Ringing

```python
N          = 16
t          = np.linspace(-1., 1., N)
y          =  1. / (1 + 25*t**2)
y_measured = y + np.random.normal(size=N)

df = pd.DataFrame( {'t': t, 'y_theoretical': y, 'y_measured': y_measured})
```

```python
t2 = t*t; t3 = t*t*t; t4=t3*t; t5=t4*t; t6=t5*t; t7=t6*t; t8=t7*t;t9=t8*t;t10=t9*t;
t11=t10*t; t12=t11*t; t13=t12*t; t14=t13*t;t15=t14*t
B  = np.stack( [np.ones(len(t)), t, t2, t3, t4, t5, t6, t7, t8, t9,
                t10,t11, t12, t13, t14, t15 ],axis=1)

def solve(B, k):
    A    = B[:,0:1+k]
    # set up the normal equation
    At_A = A.T @ A
    At_y = A.T @ y_measured

    x_star = np.linalg.solve( At_A, At_y )

    #print( 'At_A =');    print(At_A); print()
    #print( 'At_y =');    print(At_y); print()
    #print('Solution ='); print(x_star);
    return A, x_star
A,x_star = solve(B,15)
```

```python
# Compute the projections

def compute(k):
    A,x_star = solve(B,k)

    y_hat = A @ x_star
    error = y_measured - y_hat
    return plot_result( t, y, y_measured, y_hat, error)

#print( 'Norm(Error) =', np.round(np.linalg.norm(error),3))
#print( 'Least mean squares error = ', np.round(error.dot( error)/len(error),3) )
#print()
hv.DynamicMap(compute, kdims='k').redim.values(k=range(1,N)).opts("Overlay", legend_position='top')
```
