# Symbolic Computations Example

# 1. Setup

```julia
using LAlatex, LaTeXStrings
set_backend!(:sympy)
sympy = import_sympy();
```

# 2. Symbolic variables

We can use Latex to write equations in a markdown cell:<br>
Enclose the latex name of a variable, e.g., a greek letter
with \\$ signs, i.e.,  `\\$ \\alpha\\$` displays as $\alpha$

To use greek letters in variable names in a code cell,<br>
type the latex name of the variable followed by a <TAB>, e.g., type<br>
`\beta\<TAB\>`

To use symbolic variable names, they maust be declared, e.g., using `@syms`,
or the `symbols` function.

```julia
a, b, c, β = sympy.symbols("a b c beta")
```

```julia
u = sympy.symbols("u")
x = sympy.symbols("x", real=true)
y1, y2 = sympy.symbols("y1 y2", positive=true)
α = sympy.symbols("alpha", integer=true, positive=true)
l_show((u,x,y1,y2,α), separator=", ")
```

```julia
u1 = sympy.Symbol("u1", positive=true)
u2 = sympy.Symbol("u2", positive=true)

sympy.solve(sympy.Eq(u1 + u2, 0), (u1, u2), dict=true)
```

```julia
sol = sympy.solve(u1 - u2)
for s in sol println(s) end
```

```julia
z = [sympy.symbols("z_$(i)", real=true) for i in 1:5]
println("A vector of variables z:")
l_show(sympy.Matrix(z), " contains ", z[1])
```

# 3. Matrix of Symbolic and Numeric Variables and Expressions

Now that we have declared the variables, we can use the in computations.

Here, we have symbolic variables in a matrix:

```julia
A = [ α^2                     sympy.pi + u1               u2;
      x + 3*sympy.exp(-x)     2*x/sympy.sqrt(x^2 + sympy.Rational(1,5))  y1^(y2+2)
]
l_show("A = ", A)
```

We can now do computations. E.g., `A'A` computes the matrix product $A^T A$

```julia
A_sym = sympy.Matrix(A)
AtA = A_sym.T * A_sym
display( l_show( L"A^T A = ", AtA,))
println()
l_show(L"\quad", " contains term at (0 based index) [2,1] = ", AtA[2,1],)
```

```julia
# The browser interface does not handle rendering this messy matrix,
#   here we display each entry separately
for i in AtA
    display(l_show(i))
end
```

# 4. Substitutions and Solutions

```julia
l_show("A =", A_sym,L",\quad \text{set } x=0:\quad",  A_sym.subs(Dict(x => 0)),
       L",\quad \text{set } x=0,\; \alpha=1:\quad", A_sym.subs(Dict(x => 0, α => 1)))
```

```julia
eq = x^2+4x+1
sol = sympy.solve(eq)
l_show( eq, "= 0 has solutions ",   l_show(set(sol...; separator=", ")))
```
