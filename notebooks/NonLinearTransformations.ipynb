# 1. Effect of Linear and NonLinear Transformations

We will use $T: \mathbb{R}^2 \rightarrow \mathbb{R}^2$ to be able to easily display the results

```python
%%julia
num_samples=1000
data = 2*(rand( num_samples, 2) .- 0.5);   # data points in [-1,1] × [-1,1]
```

## 1.1 Random Linear Transform

```python
hv.Scatter( Main.data ).opts(width=500, title=f"{Main.num_samples} observations", xlim=(-1,1), ylim=(-1,1))
```

```python
%%julia
function random_linear_transform( data )
    A = 2*rand(2,2) .- 1
    data*A'     # A*data' = (data * A')' computes the transform of A for each (x,y) column of data'
end;
```

```python
hv.Scatter( Main.random_linear_transform( Main.data )).opts(width=500, title=f"Random Linear transform of {Main.num_samples} observations") #, xlim=(-1,1), ylim=(-1,1))
```

## 1.2 Non-linear Transform of the Data

```python
%%julia
function transform( data )
    tx_data = similar(data)
    for i in 1:size(data,1)
        tx_data[i,1] = (data[i,1])^4        # Modify the function
        tx_data[i,2] = (data[i,2])^4        # Modify the function
    end
    tx_data
end
tx_data = transform( data );
```

```python
hv.Scatter( Main.transform( Main.data )).opts(width=500, title=f"Non-Linear transform of {Main.num_samples} observations") #, xlim=(-1,1), ylim=(-1,1))
```

```python
%%julia
function transform( data )
    tx_data = similar(data)
    for i in 1:size(data,1)
        # Apply a transformation to the data point data[i,:]
        tx_data[i,1] = tanh( 5*data[i,1] + 2*data[i,2])       # Modify the function
        tx_data[i,2] = data[i,2]                              # Modify the function
    end
    tx_data
end
tx_data = transform( data );
```

```python
hv.Scatter( Main.transform( Main.data )).opts(width=500, title=f"Non-Linear transform of {Main.num_samples} observations") #, xlim=(-1,1), ylim=(-1,1))
```
