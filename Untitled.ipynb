## 1.3 Aspect ratio

An $m \times n$ matrix is represented by $m \times n$ cells of constant width and constant height.

When displaying such an image, the horizontal and vertical scales should be identical.  
Otherwise the geometry of the image is distorted. For this reason image displays are usually shown with square pixels.

```python
# Maintain the aspect ratio of an Image
import xarray as xr
# convert to xarray (1 or 4 channels, associated x y coordinates)
# see:  https://github.com/marcbernot/sanpellegrinolabel/blob/main/sanpe.ipynb

def im2DataArray(im, gray_scale=False):
    inch_mm          = 25.4

    imarray          = np.array(im)
    try:
        dpi          = im.info['dpi']
    except:
        dpi          = [100,96]
    xmax_mm, ymax_mm = im.width/dpi[0]*inch_mm,im.height/dpi[1]*inch_mm
    xs               = np.linspace(0,xmax_mm,im.width)
    ys               = np.linspace(0,ymax_mm,im.height)

    if gray_scale:
        da = xr.DataArray(imarray,
                          coords = dict(y=ys,x=xs),
                          dims = ['y','x'],
                          name='value')
    else:
        da = xr.DataArray(imarray,
                          coords = dict(y=ys,x=xs,channel = list(im.getbands())),
                          dims = ['y','x','channel'],
                          name='value')
    da.attrs["gray"] = gray_scale
    return da

def show_xr(ar):
    import hvplot.xarray
    if ar.attrs["gray"]:
        return ar.hvplot.image("x", "y", "value", flip_yaxis=True, data_aspect=1)\
                 .opts(cmap="gray",colorbar=False, xaxis=None, yaxis=None)
        #return hv.Image( (ar['x'],ar['y'], ar.data), kdims=['x','y'])\
        #        .opts(cmap="gray",invert_yaxis=True, xaxis=None, yaxis=None)

    else:
        return ar.hvplot.rgb('x','y','value', bands = 'channel',flip_yaxis=True,data_aspect=1)\
                 .opts(xaxis=None,yaxis=None)
```

```python
from PIL import Image, ImageOps
img   = Image.open( 'notebooks/Figs/cat.png' )
rgb   = np.asarray(img)
print( f"Image array of dimensions {rgb.shape}")
pn.Row( raster(rgb[:,:,0]).opts(title="Red Channel",   aspect=1.3)+\
        raster(rgb[:,:,1]).opts(title="Green Channel", aspect=1.3)+\
        raster(rgb[:,:,2]).opts(title="Blue Channel",  aspect=1.3),
      )
```

```python
# We can average the channels to produce a grayscale image
gray = ImageOps.grayscale(img)
pn.Row( show_xr( im2DataArray(img)).opts(frame_width=200, title="Color Image")+\
show_xr( im2DataArray( gray, gray_scale=True) ).opts(frame_width=200, title="Grayscale Image"))
```

**Remarks:**
* the final color and gray scale images use the correct aspect ratio
* a color image is usually encoded as 3 arrays: here a red, grean and an blue channel
* a grayscale image is obtained from the colorimage by averaging the color values of a pixel from the corresponding values in each of the channels.
