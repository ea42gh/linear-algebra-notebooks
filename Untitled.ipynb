# 1. Image representations of arrays

Numerical arrays may be represented as images, providing a convenient way to visualize the structure of a matrix.

````python
A = np.random.rand(20,20)
A[8:12,8:12]=1.
A[16:,16:]=0
pn.Row( raster(A).opts( title=f"Random Array, size {A.shape}", frame_width=200, colorbar=True ),
        pn.Spacer(width=100),
        pn.pane.Markdown( """#### The entries of the matrix are mapped to gray values.  
        Large entries correspond to bright pixels<br>
         and small entries correspond to dark pixels.
        """
        )
)
````

## Aspect ratio

A matrix with (m) rows and (n) columns determines an image consisting of (m) pixel rows and (n) pixel columns.

When displaying such an image, the horizontal and vertical scales should be identical. Otherwise the geometry of the image is distorted. For this reason image displays are usually shown with square pixels.
